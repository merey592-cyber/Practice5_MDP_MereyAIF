```python

states = ["A", "B", "C", "D", "E", "S", "F", "H", "I"]
actions = ["UP", "DOWN", "LEFT", "RIGHT"]

gamma = 0.9
theta = 0.001

grid = {
    "A": (0, 0), "B": (0, 1), "C": (0, 2), "G": (0, 3),
    "D": (1, 0), "E": (1, 2), "X": (1, 3),
    "S": (2, 0), "F": (2, 1), "H": (2, 2), "I": (2, 3)
}

directions = {
    "UP": (-1, 0),
    "DOWN": (1, 0),
    "LEFT": (0, -1),
    "RIGHT": (0, 1)
}

valid_cells = set(grid.values())

def transitions(state, action):
    row, col = grid[state]
    dr, dc = directions[action]

    left = (-dc, dr)
    right = (dc, -dr)

    outcomes = [
        (0.8, (dr, dc)),
        (0.1, left),
        (0.1, right)
    ]

    results = []

    for probability, (move_r, move_c) in outcomes:
        position = (row + move_r, col + move_c)

        if position not in valid_cells:
            next_state = state
        else:
            next_state = next(
                name for name, pos in grid.items()
                if pos == position
            )

        if next_state == "G":
            reward = 10
        elif next_state == "X":
            reward = -10
        else:
            reward = -0.1

        results.append((probability, next_state, reward))

    return results


V = {s: 0.0 for s in states}

while True:
    delta = 0
    new_V = V.copy()

    for s in states:
        q_values = []

        for a in actions:
            q = 0

            for prob, next_state, reward in transitions(s, a):
                next_value = V.get(next_state, 0)
                q += prob * (reward + gamma * next_value)

            q_values.append(q)

        new_V[s] = max(q_values)
        delta = max(delta, abs(new_V[s] - V[s]))

    V = new_V

    if delta < theta:
        break


print("Final state values:")

for s in states:
    print(f"{s}: {V[s]:.3f}")


print("\nQ-values and best actions:")

for s in states:
    q_values = {}

    for a in actions:
        q = 0

        for prob, next_state, reward in transitions(s, a):
            next_value = V.get(next_state, 0)
            q += prob * (reward + gamma * next_value)

        q_values[a] = q

    best_action = max(q_values, key=q_values.get)

    print(f"\nState {s}:")
    for action, value in q_values.items():
        print(f"  {action}: {value:.3f}")
    print("  Best action:", best_action)
```
